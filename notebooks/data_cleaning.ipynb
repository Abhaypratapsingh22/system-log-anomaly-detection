### Fix Incorrect Data Types

```python
data['Response_Time_ms'] = pd.to_numeric(data['Response_Time_ms'], errors='coerce')

data['Resolution_Time_min'] = pd.to_numeric(data['Resolution_Time_min'], errors='coerce')
```

```python
data.info()
```

### Handle Missing Values

```python
data['Source'] = data['Source'].fillna("Unknown")

data['Alert_Method'] = data['Alert_Method'].fillna("Unknown")

data['Response_Time_ms'] = data['Response_Time_ms'].fillna(data['Response_Time_ms'].median())

data['Resolution_Time_min'] = data['Resolution_Time_min'].fillna(data['Resolution_Time_min'].median())

data['Affected_Services'] = data['Affected_Services'].fillna(data['Affected_Services'].median())
```

```python
data.isnull().sum()
```

### Feature Engineering

```python
data['Timestamp'] = pd.to_datetime(data['Timestamp'], errors='coerce')
```

```python
data = data.dropna(subset=['Timestamp'])
```

```python
print(data['Timestamp'].dtype)
```

```python
data['Timestamp'].isnull().sum()
```

```python
data['hour'] = data['Timestamp'].dt.hour
data['day'] = data['Timestamp'].dt.day
data['month'] = data['Timestamp'].dt.month
```

### Encode Severity

```python
severity_map = {
    "Low":1,
    "Medium":2,
    "High":3
}

data['Severity_Level'] = data['Severity'].map(severity_map)
```

### Drop Unnecessary Columns


Host_IP
Transaction_ID
User_ID

```python
data = data.drop(columns=['Host_IP'])
```

### Save Processed Dataset

```python
data.to_csv("D:/PROJECTS/system-log-anomaly-detection/data/processed/logs_cleaned.csv", index=False)
```
